###VECTOR , MATRIX AND TENSOR

```python
vector = np.array([1,2,3])
print(vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print(matrix) 

tensor = np.array([[[1,2],[3,4]],
                    [[5,6],[7,8]]])
print(tensor)                    
```

##array properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr.shape)#rows and colomn
print(arr.ndim)#gives dimension 
print(arr.size)#no. of elements
print(arr.dtype)#datatype


```

ARRAY RESHAPING

```python
arr = np.arange(12)
print(arr)
reshaped = arr.reshape((3,4))
print(reshaped)

flattened = reshaped.flatten()
print(flattened)
#ravel = returns view instead of copy
raveled = reshaped.ravel()
print(raveled)

#transpose
transpose = reshaped.T
print(transpose)
```
